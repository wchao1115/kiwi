```python
from datasets import load_dataset

# Load the 353-book Thai literature corpus
dataset = load_dataset("pythainlp/thai-tnhc2-books", split="train")

# Preview the first book's text block
print(dataset[0]["text"][:1000])
```

```python
print(len(dataset))
```

```python
print(dataset.column_names)
```

```python
dataset.save_to_disk("thai-tnhc2-books")
```

```python
import os, re

os.makedirs("thai-tnhc2-books", exist_ok=True)
for row in dataset:
    safe = re.sub(r'[\\/:*?"<>|]', "_", f"{row['id']}_{row['book']}")
    with open(f"thai-tnhc2-books/{safe}.txt", "w", encoding="utf-8") as f:
        f.write(row["text"])
```
